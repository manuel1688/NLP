# Análisis de Balanceo de Clases en un Dataset de Reseñas

Este notebook analiza la distribución de clases en un dataset de reseñas y evalúa si el conjunto está balanceado usando estadísticas simples.

## 1. Importar librerías necesarias

```python
import json
from collections import Counter
import numpy as np
```

## 2. Cargar el dataset
Puedes subir el archivo corpus_sentimiento_reviews.json desde tu equipo si no está en el entorno.

```python
# Si ejecutas en Colab, puedes subir el archivo desde tu equipo
from google.colab import files
uploaded = files.upload()

with open("corpus_sentimiento_reviews.json", "r", encoding="utf-8") as f:
    data = json.load(f)
```

## 3. Extraer etiquetas y contar frecuencia por clase

```python
labels = [item["categoria"] for item in data["reviews"]]
conteo = Counter(labels)

print("Distribución de clases:")
for clase, count in conteo.items():
    print(f"{clase}: {count}")
```

## 4. Calcular media, desviación estándar y coeficiente de variación

```python
valores = np.array(list(conteo.values()))
mean = valores.mean()
std = valores.std()
coef_var = std / mean

print(f"\nMedia: {mean}")
print(f"Desviación estándar: {std}")
print(f"Coeficiente de variación: {coef_var:.3f}")
```

## 5. Diagnóstico de balanceo
Aplicamos una regla simple usando el coeficiente de variación para diagnosticar el balanceo del dataset.

```python
if std == 0:
    print("✔ Dataset perfectamente balanceado")
elif coef_var < 0.1:
    print("✔ Bien balanceado (variación baja)")
elif coef_var < 0.3:
    print("⚠ Aceptable pero con algo de desbalance")
else:
    print("❌ Dataset desbalanceado")
```
